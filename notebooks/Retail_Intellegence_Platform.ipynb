```python

import pandas as pd
import os

data_folder = "/content/retail_data/Retail Transactions and Stocks Data"

sales_path = os.path.join(data_folder, "retail_sales_ml_apl.csv")
inventory_path = os.path.join(data_folder, "retail_inventory_ml_apl.csv")

sales = pd.read_csv(sales_path)
inventory = pd.read_csv(inventory_path)

print("=== SALES DATA ===")
print("Shape:", sales.shape)
print("\nColumns:")
print(sales.columns.tolist())
print("\nFirst 5 rows:")
display(sales.head())

print("\n=== INVENTORY DATA ===")
print("Shape:", inventory.shape)
print("\nColumns:")
print(inventory.columns.tolist())
print("\nFirst 5 rows:")
display(inventory.head())
```

```python

import pandas as pd

# Work on copies so the original data stays unchanged
sales_audit = sales.copy()
inventory_audit = inventory.copy()

# Convert sales dates and numeric fields
sales_audit["Transaction Date"] = pd.to_datetime(
    sales_audit["Transaction Date"], errors="coerce"
)

sales_numeric = [
    "Qty Sold", "Sales Amount", "Cogs", "Number of Transactions"
]

for col in sales_numeric:
    sales_audit[col] = pd.to_numeric(
        sales_audit[col], errors="coerce"
    )

# Inventory dates: handle sentinel dates such as 9999-12-31
for col in ["Start Date", "End Date"]:
    inventory_audit[col] = pd.to_datetime(
        inventory_audit[col], errors="coerce"
    )

inventory_numeric = [
    "Qty on hand", "Stocks Selling Amount", "Cost of Stocks",
    "Stock Unit Selling Price", "Stock Unit Cost Price"
]

for col in inventory_numeric:
    inventory_audit[col] = pd.to_numeric(
        inventory_audit[col], errors="coerce"
    )

print("=== DATA QUALITY SUMMARY ===")

print("\nSales missing values:")
display(
    sales_audit.isna().sum()
    .sort_values(ascending=False)
    .rename("Missing rows")
    .to_frame()
    .query("`Missing rows` > 0")
)

print("\nInventory missing values:")
display(
    inventory_audit.isna().sum()
    .sort_values(ascending=False)
    .rename("Missing rows")
    .to_frame()
    .query("`Missing rows` > 0")
)

print("\nDuplicate rows")
print("Sales:", sales_audit.duplicated().sum())
print("Inventory:", inventory_audit.duplicated().sum())

print("\nSales date range:")
print(sales_audit["Transaction Date"].min(),
      "to", sales_audit["Transaction Date"].max())

print("\nInventory date range:")
print("Start:", inventory_audit["Start Date"].min(),
      "to", inventory_audit["Start Date"].max())
print("Invalid/unparsed end dates:",
      inventory_audit["End Date"].isna().sum())

print("\nReturn indicator values:")
print(sales_audit["Is Return"].value_counts(dropna=False))

print("\nSales numeric summary:")
display(sales_audit[sales_numeric].describe().round(2))

print("\nInventory numeric summary:")
display(inventory_audit[inventory_numeric].describe().round(2))
```

```python

# 1. Inspect inventory end dates before date conversion
raw_end_dates = inventory["End Date"].astype("string").str.strip()

print("=== INVENTORY END-DATE CHECK ===")
print("9999-12-31 rows:", raw_end_dates.eq("9999-12-31").sum())
print("Blank end dates:", raw_end_dates.isna().sum())
print("\nMost common raw end dates:")
display(raw_end_dates.value_counts(dropna=False).head(10).to_frame("Rows"))

# 2. Inspect return records
print("\n=== RETURN RECORD CHECK ===")
return_rows = sales[sales["Is Return"] == 1]
non_return_rows = sales[sales["Is Return"] == 0]

print("Return rows:", len(return_rows))
print("Return rows with a missing reason:",
      return_rows["Reason of Return"].isna().sum())

print("\nReturn amount summary:")
display(
    return_rows[["Qty Sold", "Sales Amount", "Cogs",
                 "Number of Transactions"]].describe().round(2)
)

print("\nNon-return amount summary:")
display(
    non_return_rows[["Qty Sold", "Sales Amount", "Cogs",
                     "Number of Transactions"]].describe().round(2)
)

print("\nNegative amounts or quantities")
for label, df in [("Returns", return_rows),
                  ("Non-returns", non_return_rows)]:
    print(
        label,
        "| negative quantity:", (df["Qty Sold"] < 0).sum(),
        "| negative sales amount:", (df["Sales Amount"] < 0).sum(),
        "| negative COGS:", (df["Cogs"] < 0).sum()
    )
```

```python

# Compare returns and non-returns
sales["Is Return"] = pd.to_numeric(
    sales["Is Return"], errors="coerce"
)

for col in ["Qty Sold", "Sales Amount", "Cogs",
            "Number of Transactions"]:
    sales[col] = pd.to_numeric(sales[col], errors="coerce")

summary = (
    sales.groupby("Is Return", dropna=False)
    .agg(
        Records=("Product No", "size"),
        Total_Quantity=("Qty Sold", "sum"),
        Total_Sales_Amount=("Sales Amount", "sum"),
        Total_COGS=("Cogs", "sum"),
        Transaction_Count=("Number of Transactions", "sum"),
        Negative_Quantities=("Qty Sold", lambda x: (x < 0).sum()),
        Negative_Sales=("Sales Amount", lambda x: (x < 0).sum()),
        Negative_COGS=("Cogs", lambda x: (x < 0).sum())
    )
    .round(2)
)

print("=== RETURNS VS NON-RETURNS ===")
display(summary)

print("\n=== SAMPLE RETURN RECORDS ===")
display(
    sales.loc[
        sales["Is Return"] == 1,
        ["Transaction Date", "Product No", "Qty Sold",
         "Sales Amount", "Cogs", "Number of Transactions"]
    ].head(10)
)
```

```python

# STEP 1: Overall business KPIs

import pandas as pd

# Separate sales and return records
regular_sales = sales[sales["Is Return"] == 0]
return_sales = sales[sales["Is Return"] == 1]

gross_sales = regular_sales["Sales Amount"].sum()
return_value = return_sales["Sales Amount"].sum()
net_sales = sales["Sales Amount"].sum()

gross_cogs = regular_sales["Cogs"].sum()
return_cogs = return_sales["Cogs"].sum()
net_cogs = sales["Cogs"].sum()

gross_profit_before_returns = gross_sales - gross_cogs
net_gross_profit = net_sales - net_cogs

kpis = pd.DataFrame({
    "Metric": [
        "Sales before returns",
        "Recorded return value (negative)",
        "Net sales",
        "COGS before returns",
        "Return-related COGS (negative)",
        "Net COGS",
        "Gross profit before returns",
        "Gross profit after returns",
        "Gross margin after returns (%)",
        "Return value as % of sales before returns",
        "Return-record share (%)"
    ],
    "Value": [
        gross_sales,
        return_value,
        net_sales,
        gross_cogs,
        return_cogs,
        net_cogs,
        gross_profit_before_returns,
        net_gross_profit,
        (net_gross_profit / net_sales * 100) if net_sales else 0,
        (abs(return_value) / gross_sales * 100) if gross_sales else 0,
        (len(return_sales) / len(sales) * 100) if len(sales) else 0
    ]
})

print("=== OVERALL BUSINESS KPIs ===")
display(kpis.round(2))


# STEP 2: Product category performance

category_performance = (
    sales.groupby("Product Category", dropna=False)
    .agg(
        Records=("Product No", "size"),
        Units_Net=("Qty Sold", "sum"),
        Net_Sales=("Sales Amount", "sum"),
        Net_COGS=("Cogs", "sum"),
        Return_Records=("Is Return", "sum")
    )
    .reset_index()
)

category_performance["Gross_Profit"] = (
    category_performance["Net_Sales"]
    - category_performance["Net_COGS"]
)

category_performance["Gross_Margin_Pct"] = (
    category_performance["Gross_Profit"]
    / category_performance["Net_Sales"].replace(0, pd.NA)
    * 100
)

category_performance["Return_Record_Share_Pct"] = (
    category_performance["Return_Records"]
    / category_performance["Records"]
    * 100
)

category_performance = category_performance.sort_values(
    "Net_Sales", ascending=False
)

print("=== TOP 10 CATEGORIES BY NET SALES ===")
display(category_performance.head(10).round(2))

print("=== TOP 10 CATEGORIES BY GROSS PROFIT ===")
display(
    category_performance.sort_values(
        "Gross_Profit", ascending=False
    ).head(10).round(2)
)
```

```python
#Validate inventory records at the latest sales date

import pandas as pd

# Preserve the original end-date values before parsing
inventory["End Date Raw"] = inventory["End Date"].astype("string").str.strip()

inventory["Start Date"] = pd.to_datetime(
    inventory["Start Date"], errors="coerce"
)

# Treat 9999-12-31 as a special open-ended marker,
# not as a real calendar date
inventory["End Date Parsed"] = pd.to_datetime(
    inventory["End Date Raw"].mask(
        inventory["End Date Raw"] == "9999-12-31"
    ),
    errors="coerce"
)

# Convert inventory measures to numeric
for col in [
    "Qty on hand",
    "Stocks Selling Amount",
    "Cost of Stocks",
    "Stock Unit Selling Price",
    "Stock Unit Cost Price"
]:
    inventory[col] = pd.to_numeric(inventory[col], errors="coerce")

# Use the latest sales date as our analysis date
analysis_date = pd.to_datetime(sales["Transaction Date"]).max()

# A record is considered active if its start date is on/before
# the analysis date and its end date is on/after it,
# or it has the open-ended sentinel.
active_inventory = inventory[
    (inventory["Start Date"] <= analysis_date)
    & (
        (inventory["End Date Parsed"] >= analysis_date)
        | (inventory["End Date Raw"] == "9999-12-31")
    )
].copy()

print("Analysis date:", analysis_date.date())
print("Total inventory records:", len(inventory))
print("Records active on analysis date:", len(active_inventory))
print(
    "Distinct products in active inventory:",
    active_inventory["Product No"].nunique()
)
print(
    "Distinct product-store-channel combinations:",
    active_inventory[
        ["Product No", "Store", "Sales Channel"]
    ].drop_duplicates().shape[0]
)

print("\n=== ACTIVE INVENTORY BY STOCK STATUS ===")
display(
    active_inventory.groupby("Stock Status", dropna=False)
    .agg(
        Records=("Product No", "size"),
        Units_On_Hand=("Qty on hand", "sum"),
        Stock_Selling_Value=("Stocks Selling Amount", "sum"),
        Stock_Cost_Value=("Cost of Stocks", "sum")
    )
    .round(2)
)

print("\n=== INVENTORY DATE CHECK ===")
print("Earliest start date:", inventory["Start Date"].min())
print("Latest start date:", inventory["Start Date"].max())
print(
    "End dates after analysis date:",
    (inventory["End Date Parsed"] > analysis_date).sum()
)
```

```python

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

OUT = "/content/retail-inventory"
os.makedirs(OUT, exist_ok=True)

# Portfolio theme
BG = "#08111F"
PANEL = "#111D2E"
TEXT = "#F1F5F9"
MUTED = "#94A3B8"
CYAN = "#22D3EE"
TEAL = "#2DD4BF"
BLUE = "#60A5FA"
ORANGE = "#FBBF24"
RED = "#FB7185"
PURPLE = "#A78BFA"

plt.rcParams.update({
    "figure.facecolor": BG,
    "axes.facecolor": PANEL,
    "text.color": TEXT,
    "axes.labelcolor": MUTED,
    "xtick.color": MUTED,
    "ytick.color": MUTED,
    "font.family": "DejaVu Sans",
})

# Create working copies
s = sales.copy()
inv = active_inventory.copy()

# Normalize relevant columns
s["Transaction Date"] = pd.to_datetime(
    s["Transaction Date"], errors="coerce"
)
for col in ["Sales Amount", "Cogs", "Qty Sold",
            "Number of Transactions"]:
    s[col] = pd.to_numeric(s[col], errors="coerce")

# Normalize return flag, including numeric or text representations
return_flag = (
    s["Is Return"].astype("string").str.strip().str.lower()
)
s["_is_return"] = return_flag.isin(
    ["1", "true", "yes", "y"]
)

# Inventory measures
for col in ["Qty on hand", "Stocks Selling Amount",
            "Cost of Stocks"]:
    inv[col] = pd.to_numeric(inv[col], errors="coerce")

# Actual dataset calculations
valid_sales = s.dropna(subset=["Transaction Date"]).copy()

sales_before_returns = s.loc[
    ~s["_is_return"], "Sales Amount"
].sum()

recorded_return_sales = s.loc[
    s["_is_return"], "Sales Amount"
].sum()

net_sales = s["Sales Amount"].sum()

cogs_before_returns = s.loc[
    ~s["_is_return"], "Cogs"
].sum()

net_cogs = s["Cogs"].sum()
gross_profit = net_sales - net_cogs

gross_margin_pct = (
    gross_profit / net_sales * 100
    if net_sales != 0 else np.nan
)

analysis_date = valid_sales["Transaction Date"].max()

# Inventory integrity checks
combo = ["Product No", "Store", "Sales Channel"]
duplicate_combos = inv.duplicated(combo, keep=False).sum()

print("=== DATA VALIDATION ===")
print("Sales records:", f"{len(s):,}")
print("Active inventory records:", f"{len(inv):,}")
print("Latest sales date:", analysis_date.date())
print("Distinct active products:", inv["Product No"].nunique())
print("Rows involved in repeated product-store-channel combinations:",
      f"{duplicate_combos:,}")

print("\n=== VERIFIED SALES CALCULATIONS ===")
print(f"Sales excluding return rows: {sales_before_returns:,.2f}")
print(f"Sales amount on return rows: {recorded_return_sales:,.2f}")
print(f"Net sales (sum of Sales Amount): {net_sales:,.2f}")
print(f"COGS excluding return rows: {cogs_before_returns:,.2f}")
print(f"Net COGS (sum of Cogs): {net_cogs:,.2f}")
print(f"Gross profit proxy (net sales - net COGS): {gross_profit:,.2f}")
print(f"Gross margin proxy: {gross_margin_pct:.2f}%"
      if pd.notna(gross_margin_pct) else "Gross margin: N/A")

# Check whether return values are negative or positive
print("\nReturn Sales Amount sign counts:")
print(s.loc[s["_is_return"], "Sales Amount"].agg(
    ["count", "min", "max", "sum"]
))

# Reusable dashboard functions
def new_dashboard(title, subtitle):
    fig = plt.figure(figsize=(16, 9), facecolor=BG)
    fig.text(0.055, 0.945, title,
             fontsize=23, fontweight="bold", color=TEXT)
    fig.text(0.057, 0.902, subtitle,
             fontsize=10, color=MUTED)
    return fig

def add_kpi(fig, x, label, value, accent):
    fig.text(x, 0.815, label.upper(),
             fontsize=8.5, color=MUTED)
    fig.text(x, 0.755, value,
             fontsize=18, fontweight="bold", color=accent)

def add_panel(fig, rect, title):
    ax = fig.add_axes(rect)
    ax.set_facecolor(PANEL)
    ax.set_title(title, loc="left", pad=12,
                 fontsize=12, fontweight="bold", color=TEXT)
    for spine in ax.spines.values():
        spine.set_visible(False)
    ax.grid(axis="y", alpha=0.12)
    ax.set_axisbelow(True)
    return ax

def save_dashboard(fig, filename):
    fig.text(
        0.055, 0.025,
        f"RETAIL INVENTORY ANALYTICS  |  DATA THROUGH "
        f"{analysis_date:%d %b %Y}",
        fontsize=8, color=MUTED
    )
    path = os.path.join(OUT, filename)
    fig.savefig(path, dpi=160, facecolor=BG)
    plt.show()
    plt.close(fig)
    print("Saved:", path)

def compact_value(value):
    """Compact monetary-style number; does not assume a currency."""
    value = float(value)
    if abs(value) >= 1_000_000:
        return f"{value/1_000_000:,.2f}M"
    if abs(value) >= 1_000:
        return f"{value/1_000:,.1f}K"
    return f"{value:,.0f}"

print("\nDashboard setup complete.")
```

```python

fig = new_dashboard(
    "Retail Sales Performance",
    "Business KPIs | Returns | Sales type analysis"
)

add_kpi(fig, .06, "Net sales", compact_value(net_sales), CYAN)
add_kpi(fig, .29, "Recorded returns", compact_value(recorded_return_sales), RED)
add_kpi(fig, .52, "Net COGS", compact_value(net_cogs), ORANGE)
add_kpi(fig, .75, "Gross profit proxy", compact_value(gross_profit), TEAL)

# Sales reconciliation
ax1 = add_panel(
    fig, [.07, .37, .40, .29],
    "Sales excluding returns vs net sales"
)
labels = ["Sales excluding return rows", "Net sales"]
values = [sales_before_returns, net_sales]
bars = ax1.barh(labels, values, color=[BLUE, CYAN])
ax1.invert_yaxis()
ax1.xaxis.set_major_formatter(
    FuncFormatter(lambda x, _: compact_value(x))
)
ax1.bar_label(
    bars, labels=[compact_value(v) for v in values],
    padding=5, color=TEXT, fontsize=9
)
ax1.set_xlim(min(0, min(values) * 1.2),
             max(values) * 1.3 if max(values) > 0 else 1)
ax1.grid(axis="x", alpha=.12)
ax1.grid(axis="y", visible=False)

# Sales type
ax2 = add_panel(
    fig, [.54, .37, .39, .29],
    "Net sales by recorded sales type"
)
by_type = (
    s.groupby("Sales Type", dropna=False)["Sales Amount"]
    .sum()
    .sort_values()
)
type_labels = by_type.index.to_series().fillna("Unknown").astype(str)
bars = ax2.barh(
    type_labels, by_type.values,
    color=[ORANGE if v < 0 else CYAN for v in by_type.values]
)
ax2.xaxis.set_major_formatter(
    FuncFormatter(lambda x, _: compact_value(x))
)
ax2.bar_label(
    bars, labels=[compact_value(v) for v in by_type.values],
    padding=4, color=TEXT, fontsize=8
)
ax2.grid(axis="x", alpha=.12)
ax2.grid(axis="y", visible=False)

# Gross margin indicator
margin_text = (
    f"{gross_margin_pct:.1f}%"
    if pd.notna(gross_margin_pct) else "N/A"
)
fig.text(.55, .285, f"Gross margin proxy: {margin_text}",
         fontsize=11, color=TEAL, fontweight="bold")
fig.text(.55, .245,
         "Calculated as (net sales − net COGS) / net sales.",
         fontsize=8, color=MUTED)

save_dashboard(fig, "kpi-results.png")
```

```python

fig = new_dashboard(
    "Inventory Performance",
    "Active stock snapshot | Stock value | Stock cost"
)

total_units = inv["Qty on hand"].sum()
stock_selling_value = inv["Stocks Selling Amount"].sum()
stock_cost_value = inv["Cost of Stocks"].sum()
stock_value_gap = stock_selling_value - stock_cost_value

add_kpi(fig, .06, "Active inventory records", f"{len(inv):,}", CYAN)
add_kpi(fig, .29, "Units on hand", f"{total_units:,.0f}", TEAL)
add_kpi(fig, .52, "Stock selling value", compact_value(stock_selling_value), BLUE)
add_kpi(fig, .75, "Stock cost value", compact_value(stock_cost_value), ORANGE)

# Status breakdown
status = (
    inv.groupby("Stock Status", dropna=False)
    .agg(
        Records=("Product No", "size"),
        Units=("Qty on hand", "sum"),
        Selling_Value=("Stocks Selling Amount", "sum"),
        Cost_Value=("Cost of Stocks", "sum")
    )
    .reset_index()
)
status["Stock Status"] = (
    status["Stock Status"].fillna("Unknown").astype(str)
)
status = status.sort_values("Selling_Value")

ax1 = add_panel(
    fig, [.07, .34, .48, .33],
    "Stock selling value by status"
)
bars = ax1.barh(
    status["Stock Status"],
    status["Selling_Value"],
    color=[CYAN, TEAL, ORANGE, RED, PURPLE][:len(status)]
    if len(status) <= 5 else CYAN
)
ax1.xaxis.set_major_formatter(
    FuncFormatter(lambda x, _: compact_value(x))
)
ax1.bar_label(
    bars,
    labels=[compact_value(v) for v in status["Selling_Value"]],
    padding=4, color=TEXT, fontsize=8
)
ax1.grid(axis="x", alpha=.12)
ax1.grid(axis="y", visible=False)

# Stock selling value vs cost value
ax2 = add_panel(
    fig, [.62, .34, .31, .33],
    "Selling value vs cost value"
)
metrics = ["Selling value", "Cost value"]
metric_values = [stock_selling_value, stock_cost_value]
bars = ax2.bar(
    metrics, metric_values,
    color=[CYAN, ORANGE], width=.55
)
ax2.yaxis.set_major_formatter(
    FuncFormatter(lambda x, _: compact_value(x))
)
ax2.bar_label(
    bars, labels=[compact_value(v) for v in metric_values],
    padding=4, color=TEXT, fontsize=9
)
ax2.tick_params(axis="x", labelrotation=8)
ax2.grid(axis="y", alpha=.12)

fig.text(
    .07, .245,
    f"Difference between selling value and cost value: "
    f"{compact_value(stock_value_gap)}",
    fontsize=11, color=TEAL, fontweight="bold"
)
fig.text(
    .07, .205,
    f"Repeated product-store-channel rows: {duplicate_combos:,} "
    "(rows involved in duplicates; verify before aggregating)",
    fontsize=8, color=MUTED
)

save_dashboard(fig, "inventory-analysis.png")

print("\n=== STOCK STATUS SUMMARY ===")
display(status.round(2))
```
